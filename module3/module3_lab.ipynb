```python

import pandas as pd
df = pd.read_csv ("messy_sales.csv")

print(df.head(10))

print("Dataset Shape", df.shape)
print("Column Types")
print(df.dtypes)

```

After printing the preview of the dataset messy_sales, the three problems I immediately catch are: Line 6 is missing a integer inside of the zip column, line 3 and 5 are missing data in the price column, lastly the date section has 2 different types of date formats, instead of one.

```python
df.isna().sum()
```

```python
fill_value = df["price"].median()
df["price"] = df["price"].fillna(fill_value)

print(fill_value)
print(df.isna().sum())
```

```python
print(df.duplicated().sum())
df = df.drop_duplicates()

print("Dataset Shape", df.shape)
```

```python
df["zip"] = df["zip"].astype(str).str.zfill(5)
df.head(10)
```

```python
df["date"] = pd.to_datetime(df["date"], format="mixed")

print(df.shape)
print(df.dtypes)
df.head(10)
```

```python
df[df["qty"] < 0]
```

```python
df = df[df["qty"] >= 0 ]
print(df.shape)
```

```python
df.to_csv("sales_clean.csv" , index = False )
```
