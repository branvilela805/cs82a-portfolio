```python
import pandas as pd

df = pd.read_csv("sales_clean.csv")

print(df.shape)
df.head()
```

```python
print(df["price"].mean())
print(df["price"].median())
print(df["price"].std())

print(df["qty"].mean())
print(df["qty"].median())
print(df["qty"].std())
```
