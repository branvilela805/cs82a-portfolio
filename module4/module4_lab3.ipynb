By using the function df["Y"].mean() I was able to find the mean of all the best sales price provided by the dataset. I also used df["X"].mean() to find the mean of all the listed prices provided by that same dataset. By dividing Y by X and multiplying by the price of a pickup truck whose listing price is provided by the book, 25.2, I was able to come up with my predicted best sales price of 22.07587773075076. Which when rounded is $22,076.
