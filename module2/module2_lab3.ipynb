```python
Temperatures = [ 50, 58, 68, 75, 85, 98, 100]

for fahrenheit in Temperatures:
    if fahrenheit < 52:
        weather = "Freezing"
    elif fahrenheit < 60:
        weather = "Cold"
    elif fahrenheit < 75:
        weather = "Mild"
    elif fahrenheit < 90:
        weather = "Warm" 
    else:
        weather = "Hot"


    print(fahrenheit, "is", weather)

```

I live in Southern California, these temperature cutoffs represent my humnle opinion. Any temperature under 52° F is freezing. Any temperature under 60° F is Cold. Any temperature under 75° F is mild. Any temperature under 90° F is Warm. Lastly, everything over 90 is considered hot.
